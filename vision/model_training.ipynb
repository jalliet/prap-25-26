```python
import ollama
from pathlib import Path

def detect_cards(image_path: str = "image.png") -> str:
    response = ollama.chat(
        model="llava:13b",
        messages=[
            {
                "role": "user",
                "content": "What cards are in this image? List each card rank and suit, and provide a count of each unique card type.",
                "images": [image_path],
            },
        ],
    )

    return response["message"]["content"]

if __name__ == "__main__":
    image_path = "image.png"
    result = detect_cards(image_path)
    print("Card detection result:")
    print(result)
```
